## 1. Load Cleaned Datasets
We load the clean datasets prepared in Phase 1 (with historical team name standardization).

```python
matches_path = '../data/processed/matches_clean.csv'
deliveries_path = '../data/processed/deliveries_clean.csv'

df_m = pd.read_csv(matches_path)
df_d = pd.read_csv(deliveries_path)

print(f"Cleaned Matches Shape: {df_m.shape}")
print(f"Cleaned Deliveries Shape: {df_d.shape}")
```

## 2. Match-Level EDA
Inspecting match distribution per season, result types, and top venues.

```python
m_per_season = df_m['season'].value_counts().sort_index()

plt.figure(figsize=(12, 5))
sns.barplot(x=m_per_season.index, y=m_per_season.values, color='#008080')
plt.title('IPL Matches Played per Season (2008-2024)', fontweight='bold')
plt.xlabel('Season')
plt.ylabel('Matches')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
